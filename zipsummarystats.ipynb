```python
#Ian Saltzman 11/10/26 Data Summary Stats for Zip Codes of Students and Congressional Districts

from google.colab import drive
import pandas as pd

# Step 1: Mount Google Drive
drive.mount('/content/drive')

# Step 2: Define file path in "Colab Notebooks" folder
file_path = '/content/drive/MyDrive/Colab Notebooks/studentcongress.csv'

# Step 3: Read CSV into pandas DataFrame
df = pd.read_csv(file_path)

# Step 4: Verify import
print("Successfully loaded dataset:")
print(f"Total Rows: {df.shape[0]}, Total Columns: {df.shape[1]}\n")
print(df.head())

# Group by Congressional District and count unique student IDs
district_counts = df.groupby('CongDistr')['State_StudentNumber'].nunique().reset_index()
district_counts.columns = ['Congressional District', 'Unique Student Count']

# Sort in descending order
district_counts = district_counts.sort_values(by='Unique Student Count', ascending=False).reset_index(drop=True)

display(district_counts)





```
